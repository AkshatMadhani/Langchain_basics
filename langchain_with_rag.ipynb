```python
!pip install pypdf
from langchain_community.document_loaders import PyPDFLoader
loader= PyPDFLoader('vidit-mehra-updated.pdf')
docs = loader.load()
```

```python
docs[0].page_content
```

```python
#transform docs in to smaller chunks
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=0)
texter=text_splitter.split_documents(docs)
texter[:2]
```

```python
from google.colab import userdata
google_api_key=userdata.get('google_api_key')
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain.vectorstores import FAISS
db = FAISS.from_documents(texter,GoogleGenerativeAIEmbeddings(documents=texter,model="models/embedding-001",google_api_key=google_api_key))
```

```python
query = "emergency actions"
retireved_results=db.similarity_search(query)
print(retireved_results[0].page_content)
```
